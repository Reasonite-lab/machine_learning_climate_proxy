```python
import pandas as pd
import numpy as np
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)

# ===============================
# 读取数据
# ===============================
# 含传统CLAMP结果的文件
df_true = pd.read_excel(
    "Global378+Tibet15+Fossil73_更正.xlsx",
    sheet_name="fossils"
)

# 机器学习预测结果
df_pred = pd.read_csv(
    "fossil_paleoclimate_predictions.csv"
)

# 七个气候变量
variables = [
    'MAT',
    'WMMT',
    'CMMT',
    'GROWSEAS',
    'RH',
    'SH',
    'ENTHAL'
]

results = []

# ===============================
# 计算指标
# ===============================
for var in variables:

    y_true = df_true[var]
    y_pred = df_pred[var]

    r2 = r2_score(y_true, y_pred)

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    bias = np.mean(
        y_pred - y_true
    )

    results.append({
        "Variable": var,
        "R²": round(r2, 3),
        "RMSE": round(rmse, 3),
        "MAE": round(mae, 3),
        "Bias": round(bias, 3)
    })

# ===============================
# 输出表格
# ===============================
table3 = pd.DataFrame(results)

print(table3)

# 保存为Excel
table3.to_excel(
    "Table3_ML_vs_Traditional_CLAMP.xlsx",
    index=False
)

print("\nTable 3 saved successfully.")
```
