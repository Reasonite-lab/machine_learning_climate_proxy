```python
import joblib
import pandas as pd

# ==========================
# 读取化石数据
# ==========================
file_path = "Global378+Tibet15+Fossil73_更正.xlsx"

df_fossil = pd.read_excel(
    file_path,
    sheet_name="fossils"
)

# ==========================
# 31个叶性状
# ==========================
feature_cols = [
    'Lobed', 'No Teeth', 'Teeth Regul', 'Teeth Close',
    'Teeth Round', 'Teeth Acute', 'Teeth Compo',
    'Nanophyl', 'Leptophyl', 'Leptophyl 2',
    'Microphyl 1', 'Microphyl 2', 'Microphyl 3',
    'Mesophyl 1', 'Mesophyl 2', 'Mesophyl 3',
    'Apex Emargi', 'Apex Round', 'Apex Acute',
    'Apex Attenu',
    'Base Cordat', 'Base Round', 'Base Acute',
    'L:W <1:1', 'L:W 1-2:1', 'L:W 2-3:1',
    'L:W 3-4:1', 'L:W >4:1',
    'Obovate', 'Elliptic', 'Ovate'
]

X_fossil = df_fossil[feature_cols]

# ==========================
# 最佳模型
# ==========================
best_models = {
    "MAT": "MAT_Stacking.joblib",
    "WMMT": "WMMT_CatBoost.joblib",
    "CMMT": "CMMT_Stacking.joblib",
    "GROWSEAS": "GROWSEAS_CatBoost.joblib",
    "RH": "RH_Stacking.joblib",
    "SH": "SH_Stacking.joblib",
    "ENTHAL": "ENTHAL_Stacking.joblib"
}

# ==========================
# 保存结果
# ==========================
results = pd.DataFrame()

# 如果有样品名称，可以保留
if "Site" in df_fossil.columns:
    results["Site"] = df_fossil["Site"]

# ==========================
# 循环预测
# ==========================
for variable, model_path in best_models.items():

    print(f"Predicting {variable} ...")

    model = joblib.load(model_path)

    y_pred = model.predict(X_fossil)

    results[variable] = y_pred

# ==========================
# 保存结果
# ==========================
results.to_excel(
    "Fossil_Predictions_BestModels.xlsx",
    index=False
)

print("\nPrediction finished.")
print(results.head())
```
